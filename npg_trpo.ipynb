### Геометрия пространства стратегий

Шаг градиентного подъёма можно представить как задачу локальной максимизации линейного приближения целевого функционала в окрестности точки $\boldsymbol{\theta}_k$ с квадратичным регуляризатором за евклидово смещение:
$$
\boldsymbol{\theta}_{k+1} = \underset{\boldsymbol{\theta} \in \mathbb{R}^d}{\operatorname{argmax}} \left[ J(\boldsymbol{\theta}_k) + \nabla_{\boldsymbol{\theta}}^T \left. J (\boldsymbol{\theta})\right|_{\boldsymbol{\theta} = \boldsymbol{\theta}_k} (\boldsymbol{\theta} - \boldsymbol{\theta}_k) - \frac{1}{2\alpha} \|\boldsymbol{\theta} - \boldsymbol{\theta}_k\|_2^2 \right].
$$
Дифференцируя выражение в скобках по вектору $\boldsymbol{\theta}$ и приравнивая производную к нулю, получаем тождество:
$$
\nabla_{\boldsymbol{\theta}} J(\boldsymbol{\theta}_k) - \frac{1}{\alpha}(\boldsymbol{\theta} - \boldsymbol{\theta}_k) = \mathbf{0} \implies \boldsymbol{\theta}_{k+1} = \boldsymbol{\theta}_k + \alpha \nabla_{\boldsymbol{\theta}} J(\boldsymbol{\theta}_k).
$$
Квадратичное слагаемое $\|\boldsymbol{\theta} - \boldsymbol{\theta}_k\|_2^2 = \sum\limits_{i=1}^d (\theta_i - \theta_{k, i})^2$ трактует все направления в пространстве весов одинаково.

### Дивергенция Кульбака–Лейблера

Для построения устойчивого процесса оптимизации метрика расстояния должна определяться не в пространстве параметров модели $\mathbb{R}^d$, а в пространстве порождаемых распределений вероятностей. Пространство параметризованных вероятностных распределений $\{\pi_{\boldsymbol{\theta}} \mid \boldsymbol{\theta} \in \mathbb{\Theta}\}$ не является евклидовым.

В качестве естественной меры различия распределений используют **дивергенцию Кульбака–Лейблера** $D_{\mathrm{KL}}$:
$$
D_{\mathrm{KL}}(\pi_{\boldsymbol{\theta}}(\cdot \mid s) \parallel \pi_{\tilde{\boldsymbol{\theta}}}(\cdot \mid s)) = \mathbb{E}_{a \sim \pi_{\boldsymbol{\theta}}(\cdot \mid s)} \left[ \ln \frac{\pi_{\boldsymbol{\theta}}(a \mid s)}{\pi_{\tilde{\boldsymbol{\theta}}}(a \mid s)} \right].
$$
Обобщая на всё пространство состояний:
$$
\bar{D}_{\mathrm{KL}}(\pi_{\boldsymbol{\theta}} \parallel \pi_{\tilde{\boldsymbol{\theta}}}) = \mathbb{E}_{s \sim d^{\pi_{\boldsymbol{\theta}}}} \left[ D_{\mathrm{KL}}(\pi_{\boldsymbol{\theta}}(\cdot \mid s) \parallel \pi_{\tilde{\boldsymbol{\theta}}}(\cdot \mid s)) \right],
\quad
d^{\pi_{\boldsymbol{\theta}}}(s) = (1 - \gamma) \sum_{t=0}^\infty \gamma^t \mathbb{P}(S_t = s \mid s_0 \sim \rho_0, \pi_{\boldsymbol{\theta}}),
$$
Ключевые свойства дивергенции $D_{KL}(p \parallel q) = \int\limits_{-\infty}^{\infty} p(x) \ln \frac{p(x)}{q(x)} dx$:

1. **Инвариантность к параметризации**:
    $D_{\mathrm{KL}}(p \parallel q)$ не зависит от масштаба или перепараметризации весов $\boldsymbol{\theta}$, аппроксимирующих распределения $p$ и $q$
2. **Несимметричность**:
    $D_{\mathrm{KL}}(p \parallel q) \neq D_{\mathrm{KL}}(q \parallel p)$
3. **Неотрицательность**:
    $D_{\mathrm{KL}}(p \parallel q) \geqslant 0$
4. **Равенство нулю для одинаковых распределений**:
    $D_{\mathrm{KL}}(p \parallel p) = 0$
5. **Симметричность в пределе**:
    $D_{\mathrm{KL}}(p \parallel q)$ принимает вид квадратичной формы при $q \to p$

Рассмотрим поведение дивергенции Кульбака–Лейблера при малом смещении параметров $\tilde{\boldsymbol{\theta}} = \boldsymbol{\theta} + \Delta \boldsymbol{\theta}$:
$$
\bar{D}_{\mathrm{KL}}(\pi_{\boldsymbol{\theta}} \parallel \pi_{\tilde{\boldsymbol{\theta}}}) = \bar{D}_{\mathrm{KL}}(\pi_{\boldsymbol{\theta}} \parallel \pi_{\boldsymbol{\theta}}) + \left. \nabla_{\tilde{\boldsymbol{\theta}}}^T \bar{D}_{\mathrm{KL}}(\pi_{\boldsymbol{\theta}} \parallel \pi_{\tilde{\boldsymbol{\theta}}}) \right|_{\tilde{\boldsymbol{\theta}} = \boldsymbol{\theta}} \Delta \boldsymbol{\theta} + \frac{1}{2} \Delta \boldsymbol{\theta}^T \mathbf{H}_{\mathrm{KL}}(\boldsymbol{\theta}) \Delta \boldsymbol{\theta} + O(\|\Delta \boldsymbol{\theta}\|_2^3),
\quad
[\mathbf{H}_{\mathrm{KL}}(\boldsymbol{\theta})]_{ij} = \left. \frac{\partial^2 \bar{D}_{\mathrm{KL}}(\pi_{\boldsymbol{\theta}} \parallel \pi_{\tilde{\boldsymbol{\theta}}})}{\partial \tilde{\theta}_i \partial \tilde{\theta}_j} \right|_{\tilde{\boldsymbol{\theta}} = \boldsymbol{\theta}}.
$$

При $\tilde{\boldsymbol{\theta}} = \boldsymbol{\theta}$ распределения совпадают, функция достигает глобального минимума, равного нулю. Следовательно нулевой и линейный слагаемые ряда зануляются:
$$
\bar{D}_{\mathrm{KL}}(\pi_{\boldsymbol{\theta}} \parallel \pi_{\boldsymbol{\theta} + \Delta \boldsymbol{\theta}}) = \frac{1}{2} \Delta \boldsymbol{\theta}^T \mathbf{H}_{\mathrm{KL}}(\boldsymbol{\theta}) \Delta \boldsymbol{\theta} + O(\|\Delta \boldsymbol{\theta}\|_2^3).
$$


### Информационная матрица Фишера

Для скалярного параметра $\theta$ **информация Фишера** определяется как математическое ожидание квадрата производной логарифма правдоподобия:
$$
I(\theta) = \mathbb E_{x\sim p(\cdot;\theta)} \left[ \left( \frac{\partial} {\partial\theta} \ln p(x;\theta) \right)^2 \right].
$$

Информация Фишера характеризует чувствительность распределения к изменениям параметра: чем сильнее небольшое изменение $\theta$ влияет на вероятности наблюдений, тем больше информация Фишера.

В случае $\boldsymbol{\theta}\in\mathbb R^d$ обобщением информации Фишера является **информационная матрица Фишера**. Для параметрического семейства распределений $p(x;\boldsymbol{\theta})$ она определяется как
$$
\mathbf F(\boldsymbol{\theta}) = \mathbb E_{x\sim p(\cdot;\boldsymbol{\theta})} \left[
\nabla_{\boldsymbol{\theta}}
\ln p(x;\boldsymbol{\theta}) \left(
\nabla_{\boldsymbol{\theta}}
\ln p(x;\boldsymbol{\theta})
\right)^T
\right]
$$
или в поэлементной записи для компонент $i, j \in \{1, \dots, d\}$:
$$
[\mathbf F(\boldsymbol{\theta})]_{i,j} = \mathbb E_{x\sim p(\cdot;\boldsymbol{\theta})} \left[ \frac{\partial \ln p(x;\boldsymbol{\theta})}{\partial \theta_i} \frac{\partial \ln p(x;\boldsymbol{\theta})}{\partial \theta_j} \right].
$$

Свойства матрицы $\mathbf{F}(\boldsymbol{\theta})$:

1. **Симметричность**:
   $\mathbf{F}^T = \mathbf{F}$.
2. **Положительная полуопределённость**:
   $\forall \boldsymbol{v} \in \mathbb{R}^d$ выполняется $\boldsymbol{v}^T \mathbf{F} \boldsymbol{v} = \mathbb{E} \left[ (\boldsymbol{v}^T \nabla_{\boldsymbol{\theta}} \ln p(x;\boldsymbol{\theta}))^2 \right] \geqslant 0$.
3. **Геометрический смысл**:
   $\mathbf{F}(\boldsymbol{\theta})$ выступает локальным метрическим тензором, то есть квадратичная форма $\Delta \boldsymbol{\theta}^T \mathbf{F} \Delta \boldsymbol{\theta}$ измеряет чувствительность изменения распределения стратегии при смещении весов на вектор $\Delta \boldsymbol{\theta}$.

#### Теорема (Выражение информационной матрицы Фишера через матрицу Гессе KL-дивергенции)

$$
\mathbf{F}(\boldsymbol{\theta}) = \mathbf{H}_{\mathrm{KL}}(\boldsymbol{\theta}).
$$

**Доказательство**

Зафиксируем состояние $s$ и рассмотрим частные производные по компонентам приращения $\tilde{\boldsymbol{\theta}}$:

$$
D_{\mathrm{KL}}(\pi_{\boldsymbol{\theta}} \parallel \pi_{\tilde{\boldsymbol{\theta}}}) = \int \pi_{\boldsymbol{\theta}}(a \mid s) \ln \pi_{\boldsymbol{\theta}}(a \mid s) \, da - \int \pi_{\boldsymbol{\theta}}(a \mid s) \ln \pi_{\tilde{\boldsymbol{\theta}}}(a \mid s) \, da.
$$

Вычислим вторую производную по компонентам параметров $\tilde{\theta}_i$, $\tilde{\theta}_j$:

$$
\frac{\partial^2}{\partial \tilde{\theta}_i \partial \tilde{\theta}_j} D_{\mathrm{KL}}(\pi_{\boldsymbol{\theta}} \parallel \pi_{\tilde{\boldsymbol{\theta}}}) = - \int \pi_{\boldsymbol{\theta}}(a \mid s) \frac{\partial^2 \ln \pi_{\tilde{\boldsymbol{\theta}}}(a \mid s)}{\partial \tilde{\theta}_i \partial \tilde{\theta}_j} \, da.
$$

Раскроем вторую производную логарифма:

$$
\frac{\partial^2 \ln \pi_{\tilde{\boldsymbol{\theta}}}}{\partial \tilde{\theta}_i \partial \tilde{\theta}_j} = \frac{\partial}{\partial \tilde{\theta}_j} \left( \frac{1}{\pi_{\tilde{\boldsymbol{\theta}}}} \frac{\partial \pi_{\tilde{\boldsymbol{\theta}}}}{\partial \tilde{\theta}_i} \right) = \frac{1}{\pi_{\tilde{\boldsymbol{\theta}}}} \frac{\partial^2 \pi_{\tilde{\boldsymbol{\theta}}}}{\partial \tilde{\theta}_i \partial \tilde{\theta}_j} - \frac{1}{\pi_{\tilde{\boldsymbol{\theta}}}^2} \frac{\partial \pi_{\tilde{\boldsymbol{\theta}}}}{\partial \tilde{\theta}_i} \frac{\partial \pi_{\tilde{\boldsymbol{\theta}}}}{\partial \tilde{\theta}_j}.
$$

Подставим выражение в интеграл при $\tilde{\boldsymbol{\theta}} = \boldsymbol{\theta}$:

$$
\left. \frac{\partial^2 D_{\mathrm{KL}}}{\partial \tilde{\theta}_i \partial \tilde{\theta}_j} \right|_{\tilde{\boldsymbol{\theta}} = \boldsymbol{\theta}} = - \int \frac{\partial^2 \pi_{\boldsymbol{\theta}}(a \mid s)}{\partial \theta_i \partial \theta_j} \, da + \int \pi_{\boldsymbol{\theta}}(a \mid s) \left( \frac{1}{\pi_{\boldsymbol{\theta}}(a \mid s)} \frac{\partial \pi_{\boldsymbol{\theta}}(a \mid s)}{\partial \theta_i} \right) \left( \frac{1}{\pi_{\boldsymbol{\theta}}(a \mid s)} \frac{\partial \pi_{\boldsymbol{\theta}}(a \mid s)}{\partial \theta_j} \right) da.
$$

Первый интеграл равен нулю:

$$
\int \frac{\partial^2 \pi_{\boldsymbol{\theta}}(a \mid s)}{\partial \theta_i \partial \theta_j} \, da = \frac{\partial^2}{\partial \theta_i \partial \theta_j} \int \pi_{\boldsymbol{\theta}}(a \mid s) \, da = \frac{\partial^2}{\partial \theta_i \partial \theta_j} 1 = 0.
$$

Во втором слагаемом множители в скобках представляют собой производные логарифмов:

$$
\left. \frac{\partial^2 D_{\mathrm{KL}}}{\partial \tilde{\theta}_i \partial \tilde{\theta}_j} \right|_{\tilde{\boldsymbol{\theta}} = \boldsymbol{\theta}} = \mathbb{E}_{a \sim \pi_{\boldsymbol{\theta}}(\cdot \mid s)} \left[ \frac{\partial \ln \pi_{\boldsymbol{\theta}}(a \mid s)}{\partial \theta_i} \frac{\partial \ln \pi_{\boldsymbol{\theta}}(a \mid s)}{\partial \theta_j} \right].
$$

Усреднение по распределению состояний $s \sim d^{\pi_{\boldsymbol{\theta}}}$ завершает доказательство.

### Natural Policy Gradient (NPG)

Задачу оптимизации:
$$
\Delta J = J(\boldsymbol{\theta}_k + \Delta \boldsymbol{\theta}) - J(\boldsymbol{\theta}_k) \approx \nabla_{\boldsymbol{\theta}} J^T(\boldsymbol{\theta}_k)\Delta \boldsymbol{\theta}, \quad
\Delta \boldsymbol{\theta}^* = \underset{\Delta \boldsymbol{\theta} \in \mathbb{R}^d}{\operatorname{argmax}}  \left[ \nabla_{\boldsymbol{\theta}} J^T(\boldsymbol{\theta}_k) \Delta \boldsymbol{\theta} \right], \quad \frac{1}{2} \Delta \boldsymbol{\theta}^T \mathbf{F}(\boldsymbol{\theta}_k) \Delta \boldsymbol{\theta} \leqslant \varepsilon.
$$
Метод множителей Лагранжа:
$$
\mathcal{L}(\Delta \boldsymbol{\theta}, \lambda) = \nabla_{\boldsymbol{\theta}} J ^T(\boldsymbol{\theta}) \Delta \boldsymbol{\theta} - \lambda \left( \frac{1}{2} \Delta \boldsymbol{\theta}^T \mathbf{F}(\boldsymbol{\theta}) \Delta \boldsymbol{\theta} - \varepsilon \right).
$$
$$
\nabla_{\Delta \boldsymbol{\theta}} \mathcal{L} = \nabla_{\boldsymbol{\theta}} J(\boldsymbol{\theta}) - \lambda \mathbf{F}(\boldsymbol{\theta}) \Delta \boldsymbol{\theta} = \boldsymbol{0} \implies \Delta \boldsymbol{\theta}^* = \frac{1}{\lambda} \mathbf{F}^{-1}(\boldsymbol{\theta}) \nabla_{\boldsymbol{\theta}} J(\boldsymbol{\theta}).
$$
Подставляя решение в ограничение $\Delta {\boldsymbol{\theta}^*}^T \mathbf{F} \Delta \boldsymbol{\theta}^* /2= \varepsilon$, находим множитель Лагранжа $\lambda$:
$$
\frac{1}{2 \lambda^2} \nabla_{\boldsymbol{\theta}} J^T \mathbf{F}^{-1} \mathbf{F} \mathbf{F}^{-1} \nabla_{\boldsymbol{\theta}} J = \varepsilon \implies \lambda = \sqrt{\frac{\nabla_{\boldsymbol{\theta}} J^T \mathbf{F}^{-1} \nabla_{\boldsymbol{\theta}} J}{2\varepsilon}}.
$$
Вектор $\widetilde{\nabla}_{\boldsymbol{\theta}} J = \mathbf{F}^{-1} \nabla_{\boldsymbol{\theta}} J$ называется **естественным градиентом стратегии** (**Natural Policy Gradient**, Kakade, 2001).

Размерность пространства параметров $d$ обычно составляет от $10^5$ до $10^7$ и может быть значительно больше для сложных архитектур.

* Матрица Фишера имеет размерность $d \times d$. При $d = 10^6$ её явное представление в памяти требует $10^{12}$ вещественных чисел (около 4 терабайт оперативной памяти). С учетом симметричности около 2 терабайт что не меняет порядок.
* Прямое обращение матрицы алгоритмами Гаусса требует $O(d^3)$ операций.
* В задачах обучения с подкреплением матрица Фишера, как правило, не вычисляется аналитически и должна оцениваться по выборке траекторий.

### Лемма о разности функционалов качества стратегий (Performance Difference Lemma)

Разность целевых функционалов базовой и новой стратегий $\pi$ и $\tilde{\pi}$ выражается через математическое ожидание накопленной награды новой стратегии $\tilde{\pi}$:
$$
J(\tilde{\pi}) = J(\pi) + \mathbb{E}_{\tau \sim \tilde{\pi}} \left[ \sum_{t=0}^{\infty} \gamma^t A^\pi(s_t, a_t) \right] = J(\pi) + \frac{1}{1 - \gamma} \mathbb{E}_{s \sim d^{\tilde{\pi}}, \, a \sim \tilde{\pi}} \left[ A^\pi(s, a) \right].
$$

**Доказательство**  
Рассмотрим преимущество действия базовой стратегии вдоль шагов траектории, генерируемой новой стратегией. По определению функции преимущества:
$$
A^\pi(s_t, a_t) = \mathbb{E}_{s_{t+1} \sim p} \left[ r_t + \gamma V^\pi(s_{t+1}) \;\middle|\; s_t, a_t \right] - V^\pi(s_t).
$$
Просуммируем обе части равенства по времени $t \in [0, \infty)$ с дисконтирующим множителем $\gamma^t$ под знаком математического ожидания по траектории $\tau \sim \tilde{\pi}$:
$$
\begin{aligned}
\mathbb{E}_{\tau \sim \tilde{\pi}} \left[ \sum_{t=0}^{\infty} \gamma^t A^\pi(s_t, a_t) \right] &= \mathbb{E}_{\tau \sim \tilde{\pi}} \left[ \sum_{t=0}^{\infty} \gamma^t \left( r_t + \gamma V^\pi(s_{t+1}) - V^\pi(s_t) \right) \right] \\
&= \mathbb{E}_{\tau \sim \tilde{\pi}} \left[ \sum_{t=0}^{\infty} \gamma^t r_t \right] + \mathbb{E}_{\tau \sim \tilde{\pi}}\left[\sum_{t=0}^{\infty} \gamma^t \left( \gamma V^\pi(s_{t+1}) - V^\pi(s_t) \right) \right].
\end{aligned}
$$
Вторая сумма в правой части упрощается:
$$
\begin{aligned}
\sum_{t=0}^{\infty} \left( \gamma^{t+1} V^\pi(s_{t+1}) - \gamma^t V^\pi(s_t) \right) &= \left(\gamma V^\pi(s_1) - V^\pi(s_0)\right) + \left(\gamma^2 V^\pi(s_2) - \gamma V^\pi(s_1)\right) + \dots \\
&= -V^\pi(s_0) + \lim_{T \to \infty} \gamma^T V^\pi(s_T) = -V^\pi(s_0).
\end{aligned}
$$
Подставляя результат обратно получим:
$$
\mathbb{E}_{\tau \sim \tilde{\pi}} \left[ \sum_{t=0}^{\infty} \gamma^t A^\pi(s_t, a_t) \right] = \mathbb{E}_{\tau \sim \tilde{\pi}} \left[ \sum_{t=0}^{\infty} \gamma^t r_t \right] - \mathbb{E}_{s_0 \sim \rho_0} \left[ V^\pi(s_0) \right].
$$
По определению целевого функционала:
$$
\mathbb{E}_{\tau \sim \tilde{\pi}} \left[ \sum_{t=0}^{\infty} \gamma^t r_t \right] = J(\tilde{\pi}), \quad \mathbb{E}_{s_0 \sim \rho_0} \left[ V^\pi(s_0) \right] = J(\pi),
$$
что доказывает первое равенство. Переходя от суммирования по шагам к дисконтированному распределению состояний новой стратегии
$$
d^{\tilde{\pi}}(s) = (1 - \gamma) \sum_{t=0}^\infty \gamma^t \mathbb{P}(S_t = s \mid s_0 \sim \rho_0, \tilde{\pi}),
$$
получаем представление:
$$
\begin{aligned}
\mathbb{E}_{\tau \sim \tilde{\pi}} \left[ \sum_{t=0}^{\infty} \gamma^t A^\pi(s_t,a_t) \right] &= \sum_{s \in \mathcal S} \left( \sum_{t=0}^{\infty} \gamma^t p^{\tilde{\pi}}(s) \right) \sum_{a \in \mathcal A} \tilde{\pi}(a \mid s) A^\pi(s,a) \\
&= \frac{1}{1-\gamma} \sum_{s \in \mathcal S} d^{\tilde{\pi}}(s) \sum_{a \in \mathcal A} \tilde{\pi}(a \mid s) A^\pi(s,a) \\
&= \frac{1}{1-\gamma} \mathbb{E}_{s \sim d^{\tilde{\pi}},\, a \sim \tilde{\pi}} \left[ A^\pi(s,a) \right].
\end{aligned}
$$
Отсюда следует более компактная форма записи утверждения леммы:
$$
J(\tilde{\pi}) = J(\pi) + \frac{1}{1 - \gamma} \mathbb{E}_{s \sim d^{\tilde{\pi}}, \, a \sim \tilde{\pi}} \left[ A^\pi(s, a) \right].
$$

### Суррогатная целевая функция

**Локальная суррогатная целевая функция** $L_\pi(\tilde{\pi})$: неизвестное распределение состояний $d^{\tilde{\pi}}$ заменяется известным распределением текущей стратегии $d^\pi$:
$$
L_\pi(\tilde{\pi}) = J(\pi) + \frac{1}{1 - \gamma} \mathbb{E}_{s \sim d^\pi, \, a \sim \tilde{\pi}} \left[ A^\pi(s, a) \right].
$$
Распределение $a \sim \tilde{\pi}$ заменяется на распределение $a \sim \pi$:
$$
\mathbb{E}_{a \sim \tilde{\pi}} \left[ A^\pi(s, a) \right] = \sum_{a \in \mathcal{A}} \tilde{\pi}(a \mid s) A^\pi(s, a) = \sum_{a \in \mathcal{A}} \pi(a \mid s) \frac{\tilde{\pi}(a \mid s)}{\pi(a \mid s)} A^\pi(s, a) = \mathbb{E}_{a \sim \pi} \left[ \frac{\tilde{\pi}(a \mid s)}{\pi(a \mid s)} A^\pi(s, a) \right].
$$
Тогда суррогатная целевая функция принимает вид, полностью пригодный для эмпирической оценки по накопленному батчу старой стратегии $\pi$:
$$
L_\pi(\tilde{\pi}) = J(\pi) + \frac{1}{1 - \gamma} \mathbb{E}_{s \sim d^\pi, \, a \sim \pi} \left[ \frac{\tilde{\pi}(a \mid s)}{\pi(a \mid s)} A^\pi(s, a) \right].
$$

1. В случае равенства стратегий $\tilde{\pi} = \pi$ при учете $\mathbb{E}_{a \sim \pi}[A^\pi(s, a)] = 0$:
   $$
  L_\pi(\pi) = J(\pi).
  $$
2. Градиенты суррогатной целевой функции и истинного функционала по параметрам новой стратегии в точке $\tilde{\boldsymbol{\theta}} = \boldsymbol{\theta}$ совпадают. Для доказательства раскроем градиент истинного функционала, учитывая, что параметры $\tilde{\boldsymbol{\theta}}$ входят как в распределение действий, так и в распределение состояний $d^{\pi_{\tilde{\boldsymbol{\theta}}}}(s)$:
   $$
  \begin{aligned}
  \nabla_{\tilde{\boldsymbol{\theta}}} J(\pi_{\tilde{\boldsymbol{\theta}}}) &= \frac{1}{1 - \gamma} \nabla_{\tilde{\boldsymbol{\theta}}} \sum_{s \in \mathcal{S}} d^{\pi_{\tilde{\boldsymbol{\theta}}}}(s) \sum_{a \in \mathcal{A}} \pi_{\tilde{\boldsymbol{\theta}}}(a \mid s) A^{\pi_{\boldsymbol{\theta}}}(s, a) \\
  &= \frac{1}{1 - \gamma} \sum_{s \in \mathcal{S}} \nabla_{\tilde{\boldsymbol{\theta}}} d^{\pi_{\tilde{\boldsymbol{\theta}}}}(s) \left( \sum_{a \in \mathcal{A}} \pi_{\tilde{\boldsymbol{\theta}}}(a \mid s) A^{\pi_{\boldsymbol{\theta}}}(s, a) \right) \\
  &+ \frac{1}{1 - \gamma} \sum_{s \in \mathcal{S}} d^{\pi_{\tilde{\boldsymbol{\theta}}}}(s) \nabla_{\tilde{\boldsymbol{\theta}}}\left( \sum_{a \in \mathcal{A}} \pi_{\tilde{\boldsymbol{\theta}}}(a \mid s) A^{\pi_{\boldsymbol{\theta}}}(s, a) \right).
  \end{aligned}
  $$
   Вычислим значение выражения в скобках в точке $\tilde{\boldsymbol{\theta}} = \boldsymbol{\theta}$:
   $$
  \left. \left( \sum_{a \in \mathcal{A}} \pi_{\tilde{\boldsymbol{\theta}}}(a \mid s) A^{\pi_{\boldsymbol{\theta}}}(s, a) \right) \right|_{\tilde{\boldsymbol{\theta}} = \boldsymbol{\theta}} = \mathbb{E}_{a \sim \pi_{\boldsymbol{\theta}}(\cdot \mid s)} \left[ A^{\pi_{\boldsymbol{\theta}}}(s, a) \right] = 0.
  $$
   Вклад градиента от изменения распределения состояний $\nabla_{\tilde{\boldsymbol{\theta}}} d^{\pi_{\tilde{\boldsymbol{\theta}}}}(s)$ полностью зануляется:
  $$
  \left. \nabla_{\tilde{\boldsymbol{\theta}}} J(\pi_{\tilde{\boldsymbol{\theta}}}) \right|_{\tilde{\boldsymbol{\theta}} = \boldsymbol{\theta}} = \frac{1}{1 - \gamma} \sum_{s \in \mathcal{S}} d^{\pi_{\boldsymbol{\theta}}}(s) \left. \nabla_{\tilde{\boldsymbol{\theta}}} \mathbb{E}_{a \sim \pi_{\tilde{\boldsymbol{\theta}}}} \left[ A^{\pi_{\boldsymbol{\theta}}}(s, a) \right] \right|_{\tilde{\boldsymbol{\theta}} = \boldsymbol{\theta}} = \left. \nabla_{\tilde{\boldsymbol{\theta}}} L_{\pi_{\boldsymbol{\theta}}}(\pi_{\tilde{\boldsymbol{\theta}}}) \right|_{\tilde{\boldsymbol{\theta}} = \boldsymbol{\theta}}.
  $$
  Следовательно, бесконечно малый шаг градиентного подъёма по легко вычислимой суррогатной цели $L_\pi(\tilde{\pi})$ гарантированно увеличивает исходный целевой функционал $J(\tilde{\pi})$. Однако при конечном шаге это, вообще говоря, не так.

### Теорема о монотонном улучшении (Monotonic Improvement Bound, Schulman et al., 2015)

Пусть $\varepsilon = \max\limits_{s \in \mathcal{S}, \, a \in \mathcal{A}} |A^\pi(s, a)|$ — максимум модуля функции преимущества действий базовой стратегии $\pi$. Тогда для любых двух стохастических стратегий $\pi$ и $\tilde{\pi}$ справедлива нижняя оценка прироста функционала:
$$
J(\tilde{\pi}) \geqslant L_\pi(\tilde{\pi}) - C \cdot D_{\mathrm{KL}}^{\max}(\pi \parallel \tilde{\pi}),
$$
где
$$
C = \frac{2\gamma \varepsilon}{(1 - \gamma)^2}, \quad D_{\mathrm{KL}}^{\max}(\pi \parallel \tilde{\pi}) = \max\limits_{s \in \mathcal{S}} D_{\mathrm{KL}}(\pi(\cdot \mid s) \parallel \tilde{\pi}(\cdot \mid s)).
$$

**Без доказательства**

> **Миноранта и мажоранта**
> 
> В оптимизации вспомогательная функция $g(x \mid x_k)$ называется:
> 
> * **минорантой** целевой функции $f(x)$ в точке $x_k$, если:
>   $$
g(x \mid x_k) \leqslant f(x), \quad g(x_k \mid x_k) = f(x_k), \quad \forall x;
$$
> * **мажорантой**, если:
>   $$
g(x \mid x_k) \geqslant f(x), \quad g(x_k \mid x_k) = f(x_k), \quad \forall x.
$$

Неравенство монотонного улучшения позволяет использовать **алгоритм мажорантной оптимизации** (Minorize-Maximization, MM-алгоритм):

1. На каждой итерации $k$ строится вспомогательная миноранта:
   $$
   M_{\pi_k}(\pi) = L_{\pi_k}(\pi) - C \cdot D_{\mathrm{KL}}^{\max}(\pi_k \parallel \pi).
   $$
2. Новая стратегия выбирается как глобальный максимум этой миноранты:
   $$
   \pi_{k+1} = \underset{\pi}{\operatorname{argmax}} \, M_{\pi_k}(\pi).
   $$

В базовой точке миноранта касается целевой функции $M_{\pi_k}(\pi_k) = J(\pi_k)$, при любом шаге оптимизации выполняется:
$$
J(\pi_{k+1}) \geqslant M_{\pi_k}(\pi_{k+1}) \geqslant M_{\pi_k}(\pi_k) = J(\pi_k).
$$
Гарантируется монотонное неубывание целевого функционала на каждом шаге обучения. Константа $C \propto (1 - \gamma)^{-2}$ оказывается колоссальной величиной (например, при $\gamma = 0.99$ множитель порядка $10^4$), в следствие чего алгоритм совершает исчезающе малые шаги обновления.

### TRPO (Trust Region Policy Optimization)

Вместо добавления штрафа в целевую функцию с константой $C$, дивергенция Кульбака–Лейблера выносится в явное ограничение задачи оптимизации, а трудновычислимый супремум $\max\limits_{s \in \mathcal S} D_{\mathrm{KL}}$ заменяется на математическое ожидание $\bar{D}_{\mathrm{KL}}(\pi_{\boldsymbol{\theta}} \parallel \pi_{\tilde{\boldsymbol{\theta}}}) = \mathbb{E}_{s \sim d^{\pi_{\boldsymbol{\theta}}}} \left[ D_{\mathrm{KL}}(\pi_{\boldsymbol{\theta}}(\cdot \mid s) \parallel \pi_{\tilde{\boldsymbol{\theta}}}(\cdot \mid s)) \right]$:
$$
\max_{\tilde{\boldsymbol{\theta}}} L_{\pi_{\boldsymbol{\theta}}}(\pi_{\tilde{\boldsymbol{\theta}}}), \quad \bar{D}_{\mathrm{KL}}(\pi_{\boldsymbol{\theta}} \parallel \pi_{\tilde{\boldsymbol{\theta}}}) \leqslant \delta,
$$
где $\delta > 0$ — радиус доверительной области (обычно $\delta \in [0.005, 0.05]$). Опуская постоянное слагаемое $J(\pi_{\boldsymbol{\theta}})$ и множитель $1/(1-\gamma)$, эмпирическая суррогатная целевая функция TRPO на батче траекторий:
$$
L_{\pi_{\boldsymbol{\theta}}}(\pi_{\tilde{\boldsymbol{\theta}}}) = \hat{\mathbb{E}}_k \left[ \frac{\pi_{\tilde{\boldsymbol{\theta}}}(a_k \mid s_k)}{\pi_{\boldsymbol{\theta}}(a_k \mid s_k)} \hat{A}_k \right],
$$
где $\hat{A}_k$ — оценка функции преимуществ (например, вычисленная с помощью GAE), а $\hat{\mathbb{E}}_k$ — эмпирическое среднее по батчу.

Проводится **квадратично-линейная аппроксимация**. В окрестности текущей точки $\boldsymbol{\theta}$ задача аппроксимируется:

* Суррогатная функция цели $L_{\pi_{\boldsymbol{\theta}}}(\pi_{\tilde{\boldsymbol{\theta}}})$ аппроксимируется линейной частью разложения в ряд Тейлора:
  $$
  L_{\pi_{\boldsymbol{\theta}}}(\pi_{\tilde{\boldsymbol{\theta}}}) \approx L_{\pi_{\boldsymbol{\theta}}}(\pi_{\boldsymbol{\theta}}) + \boldsymbol{g}^T (\tilde{\boldsymbol{\theta}} - \boldsymbol{\theta}), \quad \boldsymbol{g} = \left. \nabla_{\tilde{\boldsymbol{\theta}}} L_{\pi_{\boldsymbol{\theta}}}(\pi_{\tilde{\boldsymbol{\theta}}}) \right|_{\tilde{\boldsymbol{\theta}} = \boldsymbol{\theta}} = \hat{\mathbb{E}}_k \left[ \left. \nabla_{\tilde{\boldsymbol{\theta}}} \ln \pi_{\tilde{\boldsymbol{\theta}}}(a_k \mid s_k) \right|_{\tilde{\boldsymbol{\theta}} = \boldsymbol{\theta}} \hat{A}_k \right].
  $$
* Ограничение дивергенции $\bar{D}_{\mathrm{KL}}$ аппроксимируется через матрицу Фишера $\mathbf{F}$:
  $$
  \bar{D}_{\mathrm{KL}}(\pi_{\boldsymbol{\theta}} \parallel \pi_{\tilde{\boldsymbol{\theta}}}) \approx \frac{1}{2} (\tilde{\boldsymbol{\theta}} - \boldsymbol{\theta})^T \mathbf{F} (\tilde{\boldsymbol{\theta}} - \boldsymbol{\theta}).
  $$

Аппроксимированная задача в квадратично-линейной форме:
$$
\max_{\Delta \boldsymbol{\theta} \in \mathbb{R}^d} \boldsymbol{g}^T \Delta \boldsymbol{\theta}, \quad \frac{1}{2} \Delta \boldsymbol{\theta}^T \mathbf{F} \Delta \boldsymbol{\theta} \leqslant \delta, \quad \Delta \boldsymbol{\theta} = \tilde{\boldsymbol{\theta}} - \boldsymbol{\theta}.
$$
Используя метод множителей Лагранжа аналогично NPG, получаем аналитическое решение:
$$
\Delta \boldsymbol{\theta}^* = \sqrt{\frac{2\delta}{\boldsymbol{g}^T \mathbf{F}^{-1} \boldsymbol{g}}} \, \mathbf{F}^{-1} \boldsymbol{g}.
$$
Это решение требует нахождения вектора $\boldsymbol{x} = \mathbf{F}^{-1} \boldsymbol{g}$, то есть решения системы линейных уравнений
$$
\mathbf{F} \boldsymbol{x} = \boldsymbol{g}.
$$

#### Метод сопряжённых градиентов (Conjugate Gradient)

Идея TRPO: для вычисления шага не требуется знать саму матрицу $\mathbf{F}$, достаточно уметь вычислять её произведение на произвольный вектор $\boldsymbol{v} \in \mathbb{R}^d$:
$$
\boldsymbol{y} = \mathbf{F} \boldsymbol{v}.
$$
Для эффективного вычисления произведения матрицы Гессе на вектор используется метод **Hessian-Vector Product** (Pearlmutter, 1994). Произведение $\mathbf{F} \boldsymbol{v}$ вычисляется двумя последовательными проходами градиента без формирования полной матрицы:
$$
\mathbf{F} \boldsymbol{v} = \left. \nabla_{\tilde{\boldsymbol{\theta}}} \left( \nabla_{\tilde{\boldsymbol{\theta}}} \bar{D}_{\mathrm{KL}}(\pi_{\boldsymbol{\theta}} \parallel \pi_{\tilde{\boldsymbol{\theta}}}) \cdot \boldsymbol{v} \right) \right|_{\tilde{\boldsymbol{\theta}} = \boldsymbol{\theta}}.
$$
Скалярное произведение в скобках сворачивает градиент в скаляр, второй проход автодифференцирования по которому возвращает искомый вектор $\mathbf{F} \boldsymbol{v}$ с вычислительной сложностью одного стандартного шага backpropagation $O(d)$.

Система линейных уравнений вида $\mathbf{F} \boldsymbol{x} = \boldsymbol{g}$ приближённо решается итерационным **методом сопряжённых градиентов (Conjugate Gradient, CG)**. Задача решения системы $\mathbf{F} \boldsymbol{x} = \boldsymbol{g}$ эквивалентна поиску минимума квадратичной функции:
$$
\varphi(\boldsymbol{x}) = \frac{1}{2} \boldsymbol{x}^T \mathbf{F} \boldsymbol{x} - \boldsymbol{g}^T \boldsymbol{x}.
$$
В отличие от обычного градиентного спуска, где последовательные шаги могут частично компенсировать прогресс предыдущих итераций, метод сопряжённых градиентов строит систему **$\mathbf{F}$-ортогональных (сопряжённых) направлений**:
$$
\boldsymbol{p}_i^T \mathbf{F} \boldsymbol{p}_j = 0, \quad i \neq j.
$$
Благодаря этому минимизация вдоль каждого направления $\boldsymbol{p}_i$​ не ухудшает результат, достигнутый на предыдущих направлениях. Каждая итерация устраняет часть ошибки в новом независимом направлении, что существенно ускоряет сходимость метода.

#### Одномерный поиск с возвратом (Backtracking Line Search)

Поскольку вектор обновления $\Delta \boldsymbol{\theta}^*$ получен на основе приближённых (разложенных в ряд Тейлора) целевой функции и ограничения, непосредственное применение шага $\tilde{\boldsymbol{\theta}} = \boldsymbol{\theta} + \Delta \boldsymbol{\theta}^*$ может нарушить ограничение $\bar{D}_{\mathrm{KL}} \leqslant \delta$ или привести к ухудшению суррогатной цели $L_{\pi_{\boldsymbol{\theta}}}(\pi_{\tilde{\boldsymbol{\theta}}})$.

Для гарантии монотонного улучшения и строгого соблюдения ограничения в алгоритм вводится процедура **одномерного поиска с возвратом** (backtracking line search). Параметры обновляются по направлению $\Delta \boldsymbol{\theta}^*$ с экспоненциально затухающим шагом:
$$
\tilde{\boldsymbol{\theta}} = \boldsymbol{\theta} + \alpha^j \Delta \boldsymbol{\theta}^*, \quad j \in \{0, 1, 2, \dots\},
$$
где $\alpha \in (0, 1)$ — коэффициент сжатия (обычно $\alpha = 0.5$). Поиск последовательно перебирает степени $j$, принимая первое значение, при котором одновременно удовлетворяются два условия:

1. Новая стратегия улучшает суррогатную целевую функцию:
   $$
   L_{\pi_{\boldsymbol{\theta}}}(\pi_{\tilde{\boldsymbol{\theta}}}) \geqslant  L_{\pi_{\boldsymbol{\theta}}}(\pi_{\boldsymbol{\theta}}) =  J(\pi_{\boldsymbol{\theta}});
   $$
2. KL-дивергенция не выходит за границы доверительной области:
   $$
  \bar{D}_{\mathrm{KL}}(\pi_{\boldsymbol{\theta}} \parallel \pi_{\tilde{\boldsymbol{\theta}}}) \leqslant \delta.
   $$

Если за заданное число шагов подходящий масштаб шага не найден, обновление полностью отвергается, и параметры остаются прежними: $\tilde{\boldsymbol{\theta}} = \boldsymbol{\theta}$.

#### Причины ограниченного практического применения TRPO

Алгоритм TRPO совершил теоретический прорыв, впервые обеспечив стабильное обучение непрерывных глубоких стратегий. Тем не менее в современной индустрии и исследовательской практике TRPO уступил место алгоритму PPO по ряду причин:

* **Высокая алгоритмическая и инженерная сложность**  
  Реализация TRPO требует вложенных циклов: метода сопряжённых градиентов, расчёта произведений матрицы Гессе на вектор через двойной граф градиентов, процедуры линейного поиска вдоль луча.
* **Несовместимость со стандартными оптимизаторами первого порядка**  
  TRPO не может работать в связке с адаптивными градиентными методами вроде Adam или RMSprop, так как шаг формируется вручную через решение системы уравнений второго порядка.
* **Высокая вычислительная стоимость**  
  Итерации сопряжённых градиентов требуют многократных прямых и обратных проходов по всей выборке данных ради одного обновления весов, что существенно замедляет цикл обучения.
* **Проблема совместной оптимизации Актора и Критика**  
  Применение TRPO к совместной архитектуре Актора и Критика с общим каркасом признаков затруднено: ограничение доверительной области распространяется только на распределение действий выдаваемых актором, в то время как функция потерь критика должна оптимизироваться стандартным градиентным спуском.
